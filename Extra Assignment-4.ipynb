```python
gen=input("Enter the Gender:")
age=int(input("Enter the Age:"))
def Elegible(gen,age):
    print("Your Gender:",gen)
    print("Your Age:",age)
    if(gen=="Male" and age >= 21):
        eligibleMsg = "Eligible"
    elif(gen=="Female" and age >= 18):
        eligibleMsg = "Eligible"
    else:
        eligibleMsg = "Not Eligible"
    return eligibleMsg

ChkEligible=Elegible(gen,age)
ChkEligible  
```

```python
def Percentage(Sub1,Sub2,Sub3,Sub4,Sub5):
    print("Subject1:",Sub1)
    print("Subject2:",Sub2)
    print("Subject3:",Sub3)
    print("Subject4:",Sub4)
    print("Subject5:",Sub5)
    total=Sub1+Sub2+Sub3+Sub4+Sub5
    print("Total:",total)
    print("Percentage:",total/5)
    
Percentage(98,87,95,95,93)
```

```python
def AreaOftriangle(Height,breadth):
    print("Height:",Height)
    print("breadth:",breadth)
    print("Area formula:(Height*breadth)/2")
    print("Area of Triangle:",(Height*breadth)/2)

def PerimeterOftriangle(Height1,Height2,breadth):    
    print("Height1:",Height1)
    print("Height2:",Height2)
    print("breadth:",breadth)
    print("Perimeter formula:Height1+Height2+breadth")
    print("Perimeter of Triangle:",Height1+Height2+breadth)
    
AreaOftriangle(32,34)
PerimeterOftriangle(2,4,4)
```
